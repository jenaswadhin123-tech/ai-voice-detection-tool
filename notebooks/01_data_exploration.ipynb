# 01 - Data Exploration

Load sample audio files, count labels, and visualize the first audio file.

```python
from pathlib import Path
import sys

project_root = Path.cwd()
if not (project_root / 'data').is_dir():
    project_root = project_root.parent
sys.path.insert(0, str(project_root))
data_path = project_root / 'data' / 'sample_audio'
import librosa
import matplotlib.pyplot as plt

audio_extensions = {'.wav', '.mp3', '.flac', '.ogg', '.m4a'}
files = [f for f in data_path.iterdir() if f.suffix.lower() in audio_extensions]
print(f'Total audio files: {len(files)}')
print(f'AI files: {sum("ai" in f.name.lower() for f in files)}')
print(f'Human files: {sum("ai" not in f.name.lower() for f in files)}')

if files:
    y, sr = librosa.load(files[0], sr=None)
    plt.figure(figsize=(10, 4))
    plt.plot(y)
    plt.title(f'Waveform: {files[0].name}')
    plt.show()
else:
    print('No audio files found. Add audio samples to data/sample_audio.')
```
