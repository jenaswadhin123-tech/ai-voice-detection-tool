# 03 - Model Training

Train the Random Forest classifier after feature extraction.

```python
from pathlib import Path
import sys

project_root = Path.cwd()
if not (project_root / 'data').is_dir():
    project_root = project_root.parent
sys.path.insert(0, str(project_root))
data_path = project_root / 'data' / 'sample_audio'
import numpy as np
from src.feature_extraction import extract_features
from src.model import train_model

files = [f for f in data_path.iterdir() if f.suffix.lower() in {'.wav', '.mp3', '.flac', '.ogg', '.m4a'}]
features = [extract_features(f) for f in files]
labels = [1 if 'ai' in f.name.lower() else 0 for f in files]

if len(features) < 2 or len(set(labels)) < 2:
    print('Training skipped: at least one AI and one Human audio sample are required.')
    model = None
    acc = None
else:
    X = np.array(features)
    y = np.array(labels)
    model, acc = train_model(X, y)
    print(f'Accuracy: {acc}')
```
