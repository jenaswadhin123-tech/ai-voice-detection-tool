# 02 - Feature Extraction

Extract MFCC and spectral features for all audio files.

```python
from pathlib import Path
import sys

project_root = Path.cwd()
if not (project_root / 'data').is_dir():
    project_root = project_root.parent
sys.path.insert(0, str(project_root))
data_path = project_root / 'data' / 'sample_audio'
from src.feature_extraction import extract_features

features = []
labels = []
files = [f for f in data_path.iterdir() if f.suffix.lower() in {'.wav', '.mp3', '.flac', '.ogg', '.m4a'}]
for file_path in files:
    features.append(extract_features(file_path))
    labels.append(1 if 'ai' in file_path.name.lower() else 0)

print(f'Extracted features from {len(features)} audio files.')
if not features:
    print('No audio files found; add samples before training the model.')
```
